```python
from rdkit import Chem
from rdkit.Chem.Draw import IPythonConsole
IPythonConsole.drawOptions.addAtomIndices = True
IPythonConsole.molSize = (400,400)
import tabs
```

```python
m1 = Chem.AddHs(Chem.MolFromSmiles("CCCCO"))
m1
```

```python
info1 = tabs.DihedralInfoFromTorsionLib(m1)
info1.indices, info1.multiplicities
```

```python
m2 = Chem.AddHs(Chem.MolFromSmiles("C(CO)CC"))
m2
```

```python
info2 = tabs.DihedralInfoFromTorsionLib(m2)
info2.indices, info2.multiplicities
```

```python
matching = m2.GetSubstructMatch(m1)
```

```python
m2matched =  Chem.RenumberAtoms(m2, matching)
```

```python
m2matched
```

```python
infoMatched = tabs.DihedralInfoFromTorsionLib(m2matched)
infoMatched.indices, infoMatched.multiplicities
```
